## 2. Vad är WSL?

**WSL** betyder **Windows Subsystem for Linux**.

Det är ett sätt att köra en riktig Linux-miljö inuti Windows, utan att behöva installera en separat Linux-dator eller dual boot.

Varför används WSL i programmeringskurser?

- Många utvecklingsverktyg är byggda för Linux först.
- Terminalkommandon, paketinstallation och serververktyg beter sig mer likt produktionsmiljöer.
- Man kan använda Windows som vanligt, men ändå jobba i Linux när man programmerar.

Viktigt att förstå:

- Windows och Linux har olika filsystem.
- I WSL ligger Linux-filer normalt under Linux-miljöns hemkatalog, exempelvis `/home/ditt_namn`.
- Windows diskar syns ofta under `/mnt/c`, `/mnt/d` och liknande.

Rekommendation: För programmeringsprojekt i WSL är det oftast bäst att lägga projektet i Linux-filsystemet, till exempel under `~/projects`, inte djupt inne i `/mnt/c/...`. Det brukar ge färre problem och bättre prestanda.

````python
%%bash
# Visar vilket operativsystem/kernel miljön kör ovanpå.
uname -a

# Visar aktuell användare.
whoami

# Visar hemkatalogen.
echo "$HOME"
````

## 6. Skapa filer och kataloger

Vanliga kommandon:

- `mkdir namn` skapar en katalog
- `mkdir -p a/b/c` skapar flera nivåer om de saknas
- `touch fil.txt` skapar en tom fil, eller uppdaterar tidsstämpeln om filen redan finns
- `echo "text" > fil.txt` skriver text till en fil och ersätter gammalt innehåll
- `echo "mer text" >> fil.txt` lägger till text sist i filen

Skillnaden mellan `>` och `>>` är viktig. `>` skriver över. `>>` lägger till.

````python
%%bash
cd linux_lab
mkdir -p anteckningar/loggar
echo "Första raden" > anteckningar/intro.txt
echo "Andra raden" >> anteckningar/intro.txt
touch anteckningar/loggar/tom_fil.txt

find . -maxdepth 3 -type f -print
````

## 7. Läsa filer

Olika kommandon passar för olika situationer:

- `cat fil.txt` skriver ut hela filen
- `head fil.txt` visar början av filen
- `tail fil.txt` visar slutet av filen
- `less fil.txt` öppnar filen i en bläddrare i terminalen

I en notebook fungerar `cat`, `head` och `tail` bäst. I en vanlig terminal är `less` mycket användbart. Tryck `q` för att avsluta `less`.

````python
%%bash
cd linux_lab
cat anteckningar/intro.txt

echo
head -n 1 anteckningar/intro.txt
tail -n 1 anteckningar/intro.txt
````

## 8. Kopiera, flytta och ta bort

Vanliga kommandon:

- `cp källa mål` kopierar en fil
- `cp -r katalog mål` kopierar en katalog rekursivt
- `mv källa mål` flyttar eller byter namn
- `rm fil` tar bort en fil
- `rm -r katalog` tar bort en katalog och dess innehåll

Var försiktig med `rm`. I terminalen finns normalt ingen papperskorg. När man lär sig är det klokt att först köra `ls` på det man tänker ta bort.

````python
%%bash
cd linux_lab

cp anteckningar/intro.txt kopia.txt
mv kopia.txt anteckningar/kopia_av_intro.txt

echo "Efter kopiering och flytt:"
find . -maxdepth 3 -type f -print

rm anteckningar/kopia_av_intro.txt

echo
echo "Efter borttagning:"
find . -maxdepth 3 -type f -print
````

## 10. Pipes: skicka resultat vidare

En pipe, `|`, skickar output från ett kommando till nästa kommando.

Det är ett av de viktigaste Linux-mönstren.

Exempel:

```bash
ls -la | head
cat fil.txt | grep ord
find . -type f | wc -l
```

Tänk på det som en kedja:

1. första kommandot producerar text
2. nästa kommando filtrerar, räknar eller ändrar texten

````python
%%bash
cd linux_lab

echo "apa" > ord.txt
echo "banan" >> ord.txt
echo "apelsin" >> ord.txt
echo "päron" >> ord.txt

echo "Rader som innehåller 'ap':"
cat ord.txt | grep ap

echo
echo "Antal filer under linux_lab:"
find . -type f | wc -l
````

## 11. `grep`, `find` och `wc`

Tre mycket användbara verktyg:

- `grep` söker i text
- `find` söker efter filer och kataloger
- `wc` räknar rader, ord och tecken

Exempel:

```bash
grep "error" app.log
find . -name "*.py"
wc -l fil.txt
```

I moderna projekt används ofta även `rg`, ripgrep, för snabb textsökning. Men `grep` finns nästan alltid installerat.

````python
%%bash
cd linux_lab

echo "INFO Startar program" > app.log
echo "ERROR Kunde inte läsa fil" >> app.log
echo "INFO Avslutar" >> app.log

echo "Sök efter ERROR:"
grep "ERROR" app.log

echo
echo "Hitta txt-filer:"
find . -name "*.txt"

echo
echo "Antal rader i loggen:"
wc -l app.log
````

## 12. Rättigheter och `sudo`

Linux har filrättigheter. När du kör `ls -l` kan en rad börja ungefär så här:

```text
-rw-r--r--
```

Det betyder ungefär:

- ägaren får läsa och skriva
- gruppen får läsa
- andra får läsa

För scripts behöver man ibland kör-rättighet:

```bash
chmod +x script.sh
```

`sudo` betyder att kommandot körs med administratörsrättigheter. Använd det bara när du förstår varför det behövs, till exempel vid installation av paket med `apt`.

Bra regel: Om en instruktion på nätet säger `sudo`, stanna upp och fråga vad kommandot ändrar.

````python
%%bash
cd linux_lab

echo 'echo "Hej från ett shell-script"' > hej.sh
ls -l hej.sh

chmod +x hej.sh
ls -l hej.sh

./hej.sh
````

## 14. Paket och installation med `apt`

I Ubuntu och många WSL-installationer används `apt` för att installera program.

Vanligt mönster:

```bash

sudo apt install paketnamn
```

`apt update` uppdaterar listan över vilka paket som finns. Det installerar inte program i sig.

`apt install` installerar paketet.

Exempel på vanliga paket i utvecklingsmiljöer:

- `git`
- `curl`
- `build-essential`
- `python3-venv`

Naturligtvis är det skillnad mellan att installera något i Windows och att installera något inne i WSL. Det är två olika miljöer.

````python
%%bash
# Den här cellen installerar inget. Den kontrollerar bara om några vanliga program finns.
for program in git curl python3 docker; do
  if command -v "$program" >/dev/null 2>&1; then
    echo "$program finns: $(command -v "$program")"
  else
    echo "$program finns inte i PATH"
  fi
done
````

## 16. Shell-script: samla kommandon

Ett shell-script är en fil med kommandon som körs i ordning.

Det är användbart när man gör samma sak flera gånger.

Ett enkelt script kan se ut så här:

```bash
#!/usr/bin/env bash
set -e

echo "Startar"
python3 main.py
echo "Klart"
```

`set -e` betyder att scriptet avbryts om ett kommando misslyckas. Det är ofta bra i automatiserade arbetsflöden.

````python
%%bash
cd linux_lab

cat > status.sh <<'EOF'
#!/usr/bin/env bash
set -e

echo "Jag kör från: $(pwd)"
echo "Antal filer här under:"
find . -type f | wc -l
EOF

chmod +x status.sh
./status.sh
````

## 17. Vad är Docker?

Docker används för att köra program i isolerade miljöer som kallas **containers**.

En vanlig förklaring:

- En **image** är som ett recept eller en mall.
- En **container** är en körande instans av en image.
- En **volym** används för data som ska finnas kvar även om containern tas bort.
- En **port** används när något inne i containern ska nås från datorn, till exempel en webbserver.

Docker löser ett klassiskt problem: "Det fungerar på min dator".

Med Docker kan projektet beskriva sin miljö tydligare: versioner, systempaket, tjänster och startkommandon.

## 18. Docker i WSL

På Windows används Docker ofta via **Docker Desktop** tillsammans med WSL.

Typiskt arbetsflöde:

1. Installera Docker Desktop i Windows.
2. Aktivera WSL-integrationen i Docker Desktop.
3. Öppna WSL-terminalen.
4. Kontrollera med `docker version` eller `docker run hello-world`.

Om Docker-kommandot inte fungerar i WSL kan orsaken vara:

- Docker Desktop körs inte
- WSL-integrationen är inte aktiverad för rätt distribution
- användaren saknar rättigheter
- Docker är installerat i Windows men inte kopplat till WSL

````python
%%bash
# Den här cellen är säker: den kontrollerar bara om docker-kommandot finns.
if command -v docker >/dev/null 2>&1; then
  docker --version
else
  echo "Docker-kommandot finns inte i den här miljön."
fi
````

## 19. Grundläggande Docker-kommandon

De här kommandona är bra att känna igen:

```bash
docker --version
```

Skillnaden mellan `docker ps` och `docker ps -a`:

- `docker ps` visar containers som kör just nu
- `docker ps -a` visar även stoppade containers

Docker laddar ofta ner images första gången ett kommando körs. Därför kan första körningen ta längre tid.

````python
%%bash
# Kör bara Docker-kommandon om Docker finns tillgängligt.
if command -v docker >/dev/null 2>&1; then
  docker ps
else
  echo "Hoppar över: Docker finns inte tillgängligt."
fi
````

## 20. Exempel: köra ett kommando i en container

Ett vanligt Docker-mönster är:

```bash
docker run IMAGE KOMMANDO
```

Exempel:

```bash
docker run --rm ubuntu:24.04 echo "Hej från Ubuntu"
```

Vad betyder detta?

- `docker run` startar en container
- `--rm` tar bort containern när kommandot är klart
- `ubuntu:24.04` är imagen
- `echo "Hej från Ubuntu"` är kommandot som körs inne i containern

Det här kan kännas abstrakt först. Poängen är att kommandot körs i containerns Linux-miljö, inte direkt i din vanliga WSL-miljö.

````python
%%bash
# Den här cellen kan ladda ner en Docker-image om Docker fungerar.
# Om du undervisar live kan du visa kommandot först och köra det bara när alla har Docker redo.
if command -v docker >/dev/null 2>&1; then
  docker run --rm ubuntu:24.04 echo "Hej från Ubuntu i en container"
else
  echo "Hoppar över: Docker finns inte tillgängligt."
fi
````

## 21. Portar: när containern kör en server

Om en container kör en webbserver måste man ofta koppla en port i datorn till en port i containern.

Exempel:

```bash
docker run --rm -p 8080:80 nginx
```

Det betyder:

- port `80` inne i containern kopplas till port `8080` på datorn
- du kan då ofta öppna `http://localhost:8080`

Mönstret är:

```bash
-p PORT_PÅ_DATORN:PORT_I_CONTAINERN
```

Vanligt misstag: Man blandar ihop vilken port som är på datorn och vilken som är inne i containern.

## 22. Volymer: dela filer med containern

Containers är tillfälliga. Om man vill att filer ska delas mellan datorn och containern använder man ofta en volym eller bind mount.

Exempel:

```bash
docker run --rm -v "$PWD":/app -w /app ubuntu:24.04 ls
```

Det betyder:

- `-v "$PWD":/app` kopplar nuvarande katalog på datorn till `/app` i containern
- `-w /app` gör `/app` till arbetskatalog inne i containern
- `ls` körs inne i containern men ser filerna från din katalog

Det här är mycket vanligt i utvecklingsmiljöer.

````python
%%bash
cd linux_lab

if command -v docker >/dev/null 2>&1; then
  docker run --rm -v "$PWD":/app -w /app ubuntu:24.04 ls -la
else
  echo "Hoppar över: Docker finns inte tillgängligt."
fi
````

## 23. Dockerfile i korthet

En `Dockerfile` beskriver hur en image byggs.

Ett minimalt exempel:

```Dockerfile
FROM python:3.12-slim
WORKDIR /app
COPY . .
CMD ["python", "main.py"]
```

Rad för rad:

- `FROM` väljer basimage
- `WORKDIR` väljer arbetskatalog inne i imagen
- `COPY` kopierar filer från projektet in i imagen
- `CMD` anger standardkommandot när containern startas

Bygga och köra:

```bash
docker build -t mitt-program .
docker run --rm mitt-program
```
