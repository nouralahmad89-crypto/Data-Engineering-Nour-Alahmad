### Vad är CI/CD?

CI = Continuous Integration

CD = Continuous Delivery / Deployment

Idén är att utvecklare hela tiden integrerar sitt arbete i det gemensamma kodprojektet.

CI betyder att integration av kod sker kontinuerligt och att man automatiskt kan kontrollera den när förändringar skickas in. CI fokuserar alltså på att integrera och kontrollera förändringen.

CD som att nya förändringar ska kunna levereras kontinuerligt, i stället för att man väntar på en stor releas. CD handlar om att få den godkända förändringen vidare mot leverans/produktion.

Det fina med CI/CD är alltså att du som utvecklare inte behöver manuellt göra hela kontrollkedjan varje gång. Du gör din förändring och skickar in den. Sedan kan systemet automatiskt kontrollera koden och, om allt är okej, föra den vidare.

Merge conflict → Git-problem att lösa.
Test failure → CI/CD-problem som måste fixas.

**Varför behöver vi så många tester?**

Det här är viktigt, eftersom hela idén med automatisk leverans bygger på att vi har tillräckligt bra tester. man behöver många olika testfall och olika nivåer av tester för att kunna fånga olika typer av fel. Testning kan visa att ett fel finns, men inte bevisa att det inte finns några fel.

Det är omöjligt att testa alla tänkbara varianter.Därför måste man välja:**Vilka testfall är viktigast?**

Bra testtäckning + Snabba tester -> Bra CI/CD

### **Olika nivåer av tester:**

1. **Unit test :** En unit test testar en liten del av programmet, ofta en enda funktion eller modul.
Unit testing handlar inte bara om att testa när allt går rätt. Vi testar också olika felaktiga inputs för att se hur den lilla delen av programmet beter sig.

2. **Integration test :** Här testar vi hur flera komponenter fungerar tillsammans. 
"Fungerar komponenterna tillsammans?"

3. **Regressionstest :** Ett regressionstest ska alltså kontrollera att en förändring i programmet inte har förstört något som fungerade tidigare.
frågan är : Har vår nya ändring förstört något som fungerade tidigare?
oavsiktliga effekter eller kvalitet försämringar efter förändringar.

4. **Funktionella tester :** Här testar man en hel funktion i systemet ur ett användar- eller verksamhetsperspektiv.

5. **Security tests :** testar säkerheten.
6. **Penetration testing :** en person försöker angripa/hacka systemet.
7. **Acceptance tests :** testar om produkten uppfyller kundens krav.

testerna i CI/CD får inte ta för lång tid. hela idén med CI är:Gör en förändring → få snabb feedback → fortsätt eller fixa felet.

**Vad är pytest?**

pytest är ett testramverk för Python.Poängen med ett sådant ramverk är att vi slipper bygga hela infrastrukturen för testning själva varje gång vi skriver ett projekt. ramverket automatiskt kan hitta testfunktioner utifrån hur de är namngivna.

en testfunktion kan börja med:`test_` Då kan pytest känna igen den som ett testfall.

Med pytest får vi ett färdigt system för att:

- hitta testfall
- köra dem
- se vilka som lyckas
- se vilka som misslyckas

**Vad är en fixture?** En fixture är ett sätt att förbereda sådant som ett test behöver innan testet körs.

**Varför använder man fixtures?** Om flera tester behöver samma typ av förberedelse vill man inte skriva samma förberedelse om och om igen. Det gör testerna mer strukturerade och minskar upprepning.

Fixture = förbereder        Test    = kontrollerar

**Vad betyder mocking?**

Mocking innebär att man under ett test ersätter en riktig del av systemet med en simulerad version.

**Varför?** Ibland vill vi testa en viss del av vår kod utan att samtidigt behöva använda alla andra delar som den normalt är beroende av.

**Varför är det användbart?**

Det kan göra tester:
- snabbare
- mer förutsägbara
- mindre beroende av externa system

Så när vi använder mocking testar vi framför allt: "Hur beter sig min kod när den här beroende komponenten ger det här svaret?"


**Vad är pre-commit?**

pre-commit är ett verktyg som kan köra automatiska kontroller innan en Git commit görs.

Vad kan den kontrollera?
- Black – formaterar Python-kod
- isort – organiserar imports
- andra automatiska kontrolle

Du ändrar kod -> git commit -> PRE-COMMIT -> Kontroller -> Commit tillåts

Det betyder att man kan fånga vissa problem innan koden ens hamnar i repot.

**Vad är en hook?** En hook är helt enkelt en kontroll eller ett kommando som kopplas till en viss Git-händelse.
- **Pre-commit** kan alltså stoppa problem lokalt innan du committar. din dator -> pre-commit -> innan commit

- **CI/CD** kan sedan göra ytterligare kontroller när ändringen har skickats till det gemensamma repot. Github/CI -> CI/CD -> efter push/ commit

#### **Black och isort**

- **Black**

Black är ett verktyg för att automatiskt formatera Python-kod. Tanken är att alla i gruppen ska få en konsekvent kodstil. python kod -> Black -> standardiserad formatering . Det betyder att du inte behöver tänka på varje liten formateringsdetalj manuellt.

- **isort**

isort fokuserar på imports i Python. Det organiserar imports så att de får en konsekvent struktur. import-> isort -> organiserade imports

Det betyder att gruppen kan få samma kodstandard automatiskt, i stället för att varje person måste komma ihåg reglerna själv.

Varför använda dem tillsammans med pre-commit? För att automatisera kodkvalitet innan commit
