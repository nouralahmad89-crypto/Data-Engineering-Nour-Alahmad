Cron -> schema -> varje minut -> python src/countdown.py -> logs/countdown.log

I filen skriver vi : * * * * * cd /Exercise1 && python src/countdown.py för att köra programmet varje minute.

**docker run -dit --name ex1-container-1b ex1-image bash** bash på slutet ersätter Dockerfilens CMD för just den containern och håller containern igång. -dit : Kör containern i bakgrunden, håll input öppen och skapa en terminal.

**docker exec -it ex1-container-1b bash** Vi gick in i containern så nu arbetade vi inne i Linux-containern. -it : Kör containern interaktivt och ge mig en terminal.
#### Vi installerade cron:

-  Först uppdaterade vi APT:s paketlista: `apt-get update`
- Sedan installerade vi ` apt-get install -y cron nano`  cron behövs för schemaläggningen och nano använde vi för att redigera crontab. -y betyder ungefär:Answer yes automatically på frågor som installationen ställer.

- `crontab -e` -e = edit  ->skapa schemat
- `crontab -l` -l betyder list, alltså visa cron-jobben. -> kontrollera schemat

- `service cron start` -> starta cron så att schemat faktiskt körs
- `service cron status` -> kontrollera om cron-servicen körs

- `ls -l logs/countdown.log`
