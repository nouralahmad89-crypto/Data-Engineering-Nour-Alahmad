### Docker :

Docker containers är ett sätt att paketera mjukvara tillsammans med drivrutiner och delar av operativsystemet.

Docker = Packa hela programmiljön i en låda så att den fungerar överallt.

Istället för att bara skicka koden, skicka hela miljön. allt i ett paket som kallas: **Container** En container är som en liten låda. I lådan finns:
- Programmet
- Python
- Bibliotek
- Inställningar

När någon öppnar lådan får de exakt samma miljö som du hade.

**"Vad är Docker Daemon?"**

Docker Daemon är bakgrundstjänsten som bygger, startar och hanterar Docker-containrar.

**Docker Image** : En Image är en färdig mall eller ritning för hur en container ska se ut.
`docker build` skapa en image.

**Docker Container** : En Container är en körande instans av en Image. Nu arbetar programmet faktiskt.
`docker run` skapa en container från imagen.

#### Vad är en Dockerfile?

När du bygger en container måste Docker veta:

Vilket operativsystem?
Vilken Python-version?
Vilka bibliotek?
Vilken kod?
Vilket program ska startas?

Allt detta skrivs i Dockerfile.

Dockerfile är alltså instruktionerna för hur Imagen ska byggas. Den är en vamligt textfil.
`FROM python:3.13` betyder: Börja bygga min Image utifrån en färdig Python-miljö .Hämta en färdig Docker Image som redan innehåller:
- Linux-miljö
- Python 3.13
- Allt Python behöver för att köra.

använder du en Image som någon annan redan byggt. Din egen Image kommer sedan byggas ovanpå den.

`RUN pip install -r requirements.txt` RUN Körs när imagen byggs. Installera alla Python-paket innan containern startas.

**Varför gör man detta i Dockerfile?** För att alla som bygger imagen får exakt samma paket.

#### Portar

Vad är en port?

När ett program startar på en dator måste det "lyssna" på ett nummer. En port är som en dörr eller ett telefonnummer.

Exempel:

```text
FastAPI → 8000
Ollama → 11434
```

Om FastAPI kör direkt på datorn:

```text
localhost:8000
```

fungerar direkt.


#### Problemet med Docker

När FastAPI körs i en container:

```text
Container
└── FastAPI
     └── Port 8000
```

så ligger porten inne i containern.

Datorn kan inte nå den automatiskt.

#### Port Mapping

Lösning:

```yaml
ports:
  - "8000:8000"
```

Docker kopplar då:

```text
Datorns port 8000
        ↓
Containerns port 8000
```

Nu fungerar:

```text
localhost:8000
```
**Varför två nummer?**

Exempel:

```yaml
ports:
  - "9000:8000"
```

betyder:

```text
localhost:9000
        ↓
Container:8000
```

Minnesregel:

```text
ports:
"UTIFRÅN : INUTI"
```

Alltså:

```yaml
9000:8000
```

=

```text
Min dator använder 9000
Containern använder 8000
```

#### Två containers med samma port

Detta fungerar:

```text
Container A
└── Port 8000

Container B
└── Port 8000
```

Varför?

För att varje container har sitt eget nätverk.

Problemet uppstår bara på datorn.

Lösning:

```yaml
Container A:
  - "8000:8000"

Container B:
  - "9000:8000"
```

Resultat:

```text
localhost:8000
↓
Container A

localhost:9000
↓
Container B
```

Båda kör fortfarande port 8000 internt.
