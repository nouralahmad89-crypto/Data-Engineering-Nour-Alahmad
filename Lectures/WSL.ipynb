# Linux & Bash Commands Glossary

  ---------------------------------------------------------------------------
  Command                 What does the command   Useful options / examples
                          do?                     
  ----------------------- ----------------------- ---------------------------
  `cd`                    Change directory.       `cd ..`, `cd ~`,
                                                  `cd folder`

  `pwd`                   Print the current       `pwd`
                          directory.              

  `ls`                    List files and folders. `ls -l`, `ls -a`, `ls -la`,
                                                  `ls -R`

  `mkdir`                 Create a new directory. `mkdir folder`,
                                                  `mkdir -p a/b`

  `touch`                 Create an empty file.   `touch file.txt`

  `mv`                    Move or rename          `mv file folder/`,
                          files/directories.      `mv old.txt new.txt`

  `cp`                    Copy files/directories. `cp file copy.txt`,
                                                  `cp -r folder backup`

  `rm`                    Remove files.           `rm file.txt`,
                                                  `rm -r folder`

  `rmdir`                 Remove an empty         `rmdir folder`
                          directory.              

  `cat`                   Display file contents.  `cat file.txt`

  `less`                  View a file page by     `less file.txt`
                          page.                   

  `head`                  Show the first lines of `head`,
                          a file.                 `head -n 5 file.txt`

  `echo`                  Print text or           `echo "Hello"`,
                          variables.              `echo $HOME`

  `wc`                    Count lines, words and  `wc`, `wc -w`, `wc -l`,
                          bytes.                  `wc -c`

  `grep`                  Search for text in      `grep "text" file.txt`,
                          files.                  `grep -i`

  `curl`                  Download data from a    `curl URL`,
                          URL or API.             `curl -o file.json URL`

  `wget`                  Download files from the `wget URL`, `wget -O file`
                          internet.               

  `chmod`                 Change file             `chmod +x script.sh`
                          permissions.            

  `chown`                 Change file owner.      `chown user:file`

  `ssh`                   Connect to another      `ssh user@host`
                          computer securely.      

  `sudo`                  Run a command as        `sudo apt-get update`
                          administrator.          

  `apt-get`               Install or update       `update`, `install`,
                          packages.               `remove`

  `ps`                    Show running processes. `ps`, `ps -ef`, `ps aux`

  `top`                   Show running processes  `top`
                          live.                   

  `kill`                  Stop a running process. `kill PID`, `kill -9 PID`

  `find`                  Search for files and    `find . -name "*.txt"`
                          folders.                

  `diff`                  Compare two files.      `diff file1 file2`,
                                                  `diff -u`

  `&&`                    Run the next command    `cmd1 && cmd2`
                          only if the previous    
                          succeeds.               

  `>`                     Write output to a file  `echo "text" > file.txt`
                          (overwrite).            

  `>>`                    Append output to a      `echo "text" >> file.txt`
                          file.                   

  `*`                     Wildcard that matches   `*.txt`, `*.json`
                          any characters.         

  `./`                    Run a program/script in `./download_pokemons.sh`
                          the current directory.  
  ---------------------------------------------------------------------------

# Commands Used in the Exercises

  Topic               Command                        Purpose
  ------------------- ------------------------------ -------------------------------
  Create folder       `mkdir Linux-training`         Create a directory
  Create file         `touch note1.txt`              Create an empty file
  Move file           `mv note3.txt ..`              Move file to parent directory
  Rename file         `mv note3.txt note_home.txt`   Rename a file
  Delete file         `rm note4.txt`                 Remove a file
  List files          `ls`, `ls -la`, `ls -R`        Show files and folders
  Show current path   `pwd`                          Display current directory
  Print text          `echo "Hello"`                 Print text
  Save text           `>`                            Overwrite a file
  Append text         `>>`                           Add text to the end of a file
  Show file           `cat file.txt`                 Display file content
  Count words         `wc -w file.txt`               Count words
  Count lines         `wc -l file.txt`               Count lines
  Count files         `ls folder \| wc -l`           Count files in a folder
  API request         `curl URL`                     Get data from an API
  Pause               `sleep 2`                      Wait 2 seconds
  Loop                `for ... do ... done`          Repeat commands
  Variable            `$pokemon`                     Read a variable value
  Script              `#!/bin/bash`                  Bash script header
  Make executable     `chmod +x script.sh`           Allow script execution
  Run script          `./script.sh`                  Execute a Bash script
