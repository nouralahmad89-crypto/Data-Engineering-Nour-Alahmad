### Vad är Data Engineering?

Data Engineering handlar om att bygga systemen som samlar in, flyttar, lagrar och förbereder data så att andra kan använda den. **Data Engineering = bygga datasystem**

Data Engineering handlar mycket om:
- pipelines
- data processing
- Docker
- cloud
- deployment
- automation

### Vad är Agila metoder?

Agila metoder handlar inte om teknik.Det handlar om: Hur människor jobbar tillsammans i projekt. **Agila metoder = hur team bygger saker tillsammans**

Målet är:
- snabbare feedback
- mindre risk
- enklare att ändra saker
- bättre teamwork

**Scrum** är en av de mest använda agila metoderna för att organisera arbete i ett team.
Tanken är att man inte planerar hela projektet från början till slut. Istället jobbar man i korta perioder som kallas sprintar.

Efter varje sprint tittar teamet på:
- Vad blev klart?
- Vad fungerade bra?
- Vad behöver förbättras?

#### **Vanliga delar i Scrum**

1. Product Backlog : En lista med allt som behöver göras.

Exempel:
- Hämta bloggar
- Dockerisera appen
- Spara data i databas
- Bygga API

2. Sprint : En kort arbetsperiod, ofta 1–4 veckor.

3. Daily Scrum : Ett kort möte (ofta 15 minuter).

Varje person svarar ungefär på:
- Vad gjorde jag igår?
- Vad ska jag göra idag?
- Finns det några hinder?

4. Sprint Review : I slutet av sprinten visar teamet vad som fungerar.

5. Sprint Retrospective : Teamet diskuterar

- Vad gick bra?
- Vad gick mindre bra?
- Vad kan förbättras?

#### **Roller i Scrum**

1. Product Owner : Bestämmer vad som är viktigast att bygga.

2. Scrum Master  : Hjälper teamet att följa Scrum-processen.

3. Development Team : De som bygger produkten.

**Så Målet är att Anpassa dig snabbt till förändringar och leverera värde steg för steg.**

Andra agila metoder:

### Vad är Data Drift?

Data Drift betyder: Den data som modellen tränades på liknar inte längre den data som kommer in idag.

**Lösning**
- Samla ny data
- Träna om modellen

**Data Lake**

En Data Lake är en plats där man lagrar data i sitt ursprungliga format. (Rådata).

Data Lake
│
├── users.csv
├── orders.json
├── logs.txt
└── images/

Man vill alltid kunna gå tillbaka till originalet - kan man alltid börja om från rådatan. Det är därför Data Lake är viktig. När data ligger i Data Lake börjar man bearbeta den.

**Data Warehouse**

När datan är färdigbearbetad flyttas den till ett Data Warehouse. Här ligger data som är redo att användas.

**Lakehouse**

Det är i princip en kombination av Data Lake och Data Warehouse. Man försöker få fördelarna från båda.

Databricks är ett känt exempel där Lakehouse-idén används mycket.

### Story Points och Velocity: 

Story Points används för att uppskatta **hur stor eller svår en uppgift är**.

Story Points mäter **inte tid**.

Istället försöker de mäta:

```text
Komplexitet
+
Arbetsmängd
+
Osäkerhet
```

#### Exempel

```text
Fixa knapp        → 1 point

Bygga API         → 3 points

Skapa pipeline    → 5 points
```

Frågan man ställer är:

> Hur stor är denna uppgift jämfört med andra uppgifter?

Inte:

> Hur många timmar tar detta?


Velocity betyder:

> Hur mycket arbete teamet brukar hinna per sprint.

#### Exempel

Sprint 1:

```text
12 Story Points
```

Sprint 2:

```text
11 Story Points
```

Sprint 3:

```text
13 Story Points
```

Teamets ungefärliga velocity blir då:

```text
≈ 12 Story Points per sprint
```

Nu kan teamet börja planera bättre.

---

**Hur de fungerar tillsammans?**

```text
Story Points = Hur stort jobbet är

Velocity = Hur mycket teamet brukar hinna
```

Backlog : En prioriterad lista över arbete som ska göras.

- **Product Backlog**

Exempel:

#### Hur prioriterar man?

Man bygger inte allt samtidigt.

Nu när vi har:

- Backlog (vad som finns att göra)
- Story Points (hur stort arbetet är)
- Velocity (hur mycket teamet brukar hinna)

…kommer nästa steg: **Sprint**

En sprint är: En kort tidsperiod där teamet bygger något färdigt. Under sprinten försöker teamet nå ett mål.

#### Hur fungerar en sprint?

Planera → Bygga → Leverera → Förbättra → Repeat

Under sprinten försöker man undvika ändra mål hela tiden.

Om man lägger till massa nytt arbete mitt i sprinten:
- Stress ökar
- Velocity blir konstig
- Sprint Goal riskerar missas

### Scrum vs Kanban:

**Scrum** Fokus:

Arbeta i sprintar , "Vi arbetar i cykler"

_Kännetecken:_
- Arbetar i fasta sprintar
- Roller finns (PO, Scrum Master osv.)
- Planering sker inför sprintar
- Leverans sker ofta per sprint

**Kanban** Fokus:

Kontinuerligt flöde , "Fortsätt flytta arbete framåt"

_Kännetecken:_
- Inga sprintar
- Arbete flyter kontinuerligt
- Begränsar hur mycket som görs samtidigt
- Fokus på snabb genomströmning

ETL-jobb, buggar, små datafixar, supportärenden → Kanban passar ofta bra

Större projekt med tydliga mål → Scrum passar ofta bättre

# User Stories

User Stories används inom Scrum och andra agila metoder för att beskriva funktioner ur användarens perspektiv. De har i stor utsträckning ersatt traditionella kravspecifikationer eftersom de är enklare att förstå, lättare att ändra och främjar kommunikation mellan utvecklare, kunder och andra intressenter.

Grundidén är att en User Story inte ska innehålla alla tekniska detaljer. Den ska beskriva vilket behov användaren har och vilket värde funktionen skapar.

En klassisk User Story skrivs enligt mallen:

As a [roll]  
I want [funktion]  
So that [värde]

Exempel:

As a customer  
I want to log in  
So that I can access my account

Alla företag använder dock inte samma format. Vissa följer mallen strikt medan andra skriver mycket enklare beskrivningar som exempelvis:

- Implementera inloggning
- Skapa sökfunktion
- Lägg till lösenordsåterställning

Det viktiga är inte formatet utan att alla förstår vilken funktion som ska utvecklas och varför.

En User Story beskriver vad användaren vill kunna göra, men inte hur lösningen ska implementeras. Tekniska beslut som databaser, API:er eller programmeringsspråk lämnas till utvecklingsteamet.

För att kunna genomföra arbetet bryts User Stories vanligtvis ner i mindre Tasks.

Exempel:

User Story:
"Användaren ska kunna logga in"

Tasks:
- Skapa databasstruktur för användare
- Implementera inloggnings-API
- Skapa inloggningssida
- Testa funktionen

När alla Tasks är klara anses User Storyn vara färdig.

**User Stories** fungerar därför som en gemensam kommunikationspunkt mellan verksamheten och utvecklingsteamet. De hjälper teamet att fokusera på användarens behov samtidigt som arbetet kan delas upp i mindre, hanterbara delar.

User Story : Beskriver en funktion eller ett behov.

Syfte : Skapa förståelse och kommunikation.

Vanlig mall : 
As a ...
I want ...
So that ...

User Story bryts ner till Tasks

User Story : Task -> Task -> Task

### Spikes och undersökande sprintar :

En Spike används när teamet saknar kunskap för att kunna planera eller utveckla.

Resultatet är:

- kunskap
- jämförelser
- rekommendationer
- tekniska beslut

inte en färdig produkt.

Det är därför många projekt börjar med en eller flera undersökande sprintar innan den riktiga utvecklingen startar.

MVP : Minsta användbara version som ger värde till användaren.
