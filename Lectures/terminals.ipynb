Docker Desktop ska normalt vara igång när du arbetar med Docker.

Du behöver inte använda Docker Desktop GUI för varje kommando. Du kan göra nästan allt från terminalen.

### **Docker Desktop + WSL Integration**

Vi aktiverade:

Docker Desktop → Settings → Resources → WSL Integration

och valde Ubuntu.

Det gör att:

Ubuntu / WSL

      ↓

Docker CLI

      ↓

Docker Desktop

      ↓

Docker Engine

Vi testade detta med:
- docker --version
- docker info 

och fick bland annat:
- Server Version: 29.7.2
- Operating System: Docker Desktop
- Kernel Version: 6.6.87.2-microsoft-standard-WSL2

# Terminaler i Windows – Git Bash vs PowerShell vs Ubuntu/WSL

## 1. Snabb jämförelse

| | Git Bash | PowerShell | Ubuntu / WSL |
|---|---|---|---|
| Plattform | Windows | Windows | Linux via Windows |
| Riktig Linux-miljö | ❌ Nej | ❌ Nej | ✅ Ja |
| WSL | ❌ Nej | ❌ Nej | ✅ Ja |
| Git | ✅ | ✅ | ✅ |
| Docker | ✅* | ✅* | ✅ |
| Python | ✅ | ✅ | ✅ |
| `sudo` | ❌ | ❌ | ✅ |
| `apt` | ❌ | ❌ | ✅ |
| Linux-kommandon | Många | Annorlunda | ✅ |
| Rekommenderad för vår Docker-kurs | ⚪ | ⚪ | 🟢 |

\* Docker fungerar om Docker CLI är tillgängligt och korrekt konfigurerat.

sudo = "jag behöver administratörsrättigheter"

apt = "installera/hantera Ubuntu-program"

- Ubuntu är ditt WSL-operativsystem.
- Debian är operativsystemet/miljön inne i Docker-containern.
